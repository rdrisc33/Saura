```python
"""
Git is a 'version control system' for your files. 

GITIGNORE
The .gitignore file tells git which files NOT to track. 
These include dependencies, logfiles, temporary files, sensitive files, like those containing API keys, and more.
Since I'm using python, that also includes __pycache__. 
pycache` is a  directory created by Python to store compiled code, so Python can skip compilation if it detects that the `.pyc` files are up-to-date.
Thankfully, we can use a VSCode extension to create a .gitignore for the python language, removing most of the work. 

# lines starting with a # are comments and are ignored
Wildcards: asterisk * matches any character and question mark ? matches a single character
Negation: Exclamation point ! includes specific files otherwise marked for ignoring 
# ignore all .log files with *.log
# But then include important.log with !important.log 
*.log 
! important.log

# To ignore pycache folder: __pycache__/ Note the / indicates a folder.
# To ignore .pyc files: *.pyc. ( tutorial has both files and folder ignored)
__pycache__/ 
*.pyc
"""
```

```python
FORKING will copy the code from the source, and make it a repo under your own github account. 
After you fork, clone the fork to your machine so you can edit the code as if its your own


FORKING: 
Fork the repo(By pressing the button on Github)
git branch myBranch
git checkout myBranch 
make changes 
git add . to stage your files 
git commit to commit your files to the repo

git push origin myBranch to push your changes onto github. If the push is successful, you'll see a new button on github to create a pull request '
origin is the remote repository on github

Typically bad ettiquette to submit a pull request without discussing it in the forum 

can keep a fork in sync with the OG by adding a remote link to the 'upstream' repo. 
git remote add upstream <url>
When things change in that repo, can use git fetch to download the changes 
git fetch upstream 

Use git rebase to add those changes on top of your existing work
git rebase upstream/master 

```

```python
How to create a file on the windows command line: 
echo "This text will be placed in file1.txt" >> file1.txt
# The 'echo' command spits back out whatever you put into it, in this case, a string. The append operator >> appends the output of the preceding command, in this case, our echo command, and redirects it to a file.
```

```python
# Create a new repository 
echo "#test2" >>README.md
git init
git add README.md
git commit -m "first commit" 
git branch -M main 
git remote add origin https://github.com/rdrisc33/test2.git
git push -u origin main

# push an existing repository
git remote add origin https://github.com/rdrisc33/test2.git
git branch -M main                                                          # -M: shortcut for --move --force. -m|--move: move/rename a branch, together with it config and reflog(referencelogs). Will rename HEAD branch if no <old-branch> provided
git push -u origin main                                                     # update branches in remote repositories from your local repository. "git push origin main" will push the local 'main' branch to the main branch on the remote named 'origin' (???)



git pull is a combo of git fetch and git merge. Git pull fetches and merges changes in the remote repository to the local repository.
```
