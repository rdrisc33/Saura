```python
# Final example from: https://www.pythonguis.com/tutorials/creating-multiple-windows/
# Keep a reference to widgets, else they are destroyed.
# Intercept signals/slots with your own functions 

import sys
from random import randint

from PyQt5.QtWidgets import (
    QApplication,
    QLabel,
    QMainWindow,
    QPushButton,
    QVBoxLayout,
    QWidget,
)


class AnotherWindow(QWidget):
    """
    This "window" is a QWidget. If it has no parent,
    it will appear as a free-floating window.
    """

    def __init__(self):
        super().__init__()
        layout = QVBoxLayout()
        self.label = QLabel("Another Window % d" % randint(0, 100))
        layout.addWidget(self.label)
        self.setLayout(layout)


class MainWindow(QMainWindow):
    def __init__(self):
        super().__init__()
        self.window1 = AnotherWindow()
        self.window2 = AnotherWindow()

        l = QVBoxLayout()
        button1 = QPushButton("Push for Window 1")
        button1.clicked.connect(
            lambda checked: self.toggle_window(self.window1)
        )
        l.addWidget(button1)

        button2 = QPushButton("Push for Window 2")
        button2.clicked.connect(
            lambda checked: self.toggle_window(self.window2)
        )
        l.addWidget(button2)

        w = QWidget()
        w.setLayout(l)
        self.setCentralWidget(w)

    def toggle_window(self, window):
        if window.isVisible():
            window.hide()

        else:
            window.show()


app = QApplication(sys.argv)
w = MainWindow()
w.show()
app.exec()
```

```python

#Filter data for user entered text
import sys
from PyQt5.QtWidgets import QApplication, QMainWindow, QLineEdit, QListWidget, QVBoxLayout, QWidget

class SearchApp(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Search Box")

        # Create a dictionary of search options
        self.data = {
            "apple": "A fruit",
            "banana": "Another fruit",
            "cherry": "A small red fruit",
            "apricot": "Yet another fruit",
            "grape": "Small and sweet"
        }

        # Create the QLineEdit for user input
        self.entry_box = QLineEdit(self)
        self.entry_box.setPlaceholderText("Type to search...")

        # Create a QListWidget to display the search results
        self.result_list = QListWidget(self)

        # Set up the layout
        layout = QVBoxLayout()
        layout.addWidget(self.entry_box)
        layout.addWidget(self.result_list)

        # Create a central widget and set layout
        container = QWidget()
        container.setLayout(layout)
        self.setCentralWidget(container)

        # Connect the input change to the update function
        self.entry_box.textChanged.connect(self.update_search_results)

    def update_search_results(self):
        # Clear the list widget
        self.result_list.clear()

        # Get the current text from the entry box
        query = self.entry_box.text().lower()

        # Filter the data based on the query
        results = [f"{attr_group_name}: {value}" for attr_group_name, value in self.data.items()
                   if query in attr_group_name.lower() or query in value.lower()]

        # Add results to the list widget
        if results:
            self.result_list.addItems(results)

if __name__ == "__main__":
    app = QApplication(sys.argv)
    window = SearchApp()
    window.show()
    sys.exit(app.exec_())
```

```python
#Simple SVG Editor
import tkinter as tk
import drawsvg as draw

class SVGEditor:
    def __init__(self, master):
        self.master = master
        master.title("Simple SVG Editor")

        self.canvas = tk.Canvas(master, width=400, height=300)
        self.canvas.pack()

        self.drawing = draw.Drawing(400, 300)

        self.canvas.bind("<Button-1>", self.start_drawing)
        self.canvas.bind("<B1-Motion>", self.draw)

    def start_drawing(self, event):
        self.x, self.y = event.x, event.y

    def draw(self, event):
        x1, y1 = self.x, self.y
        x2, y2 = event.x, event.y
        self.canvas.create_line(x1, y1, x2, y2)
        line = draw.Line(x1, y1, x2, y2, stroke='black')
        self.drawing.append(line)
        self.x, self.y = x2, y2

    def save(self):
        self.drawing.saveSvg('output.svg')

root = tk.Tk()
editor = SVGEditor(root)
tk.Button(root, text="Save", command=editor.save).pack()
root.mainloop()
```
